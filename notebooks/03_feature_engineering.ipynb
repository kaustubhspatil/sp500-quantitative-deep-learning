# 03. High-Dimensional Technical & Macro Feature Engineering
**Objective**: Build 98 multi-factor technical indicators (RSI, MACD, Bollinger Bands, ATR, OBV, Momentum) and macroeconomic context features, formulating stationary targets.

```
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_parquet("../data/processed_features.parquet")
df['Date'] = pd.to_datetime(df['Date'])

exclude_cols = ['Date', 'Ticker', 'Next_Return', 'Target']
feature_cols = [c for c in df.columns if c not in exclude_cols]

print(f"Master Dataset Shape: {df.shape}")
print(f"Engineered Input Features: {len(feature_cols)}")
```

## 1. Directional Target Distribution

```
target_counts = df['Target'].value_counts(normalize=True)
print("Target Class Balance:")
print(f"  Down / Flat (0): {target_counts.get(0, 0)*100:.2f}%")
print(f"  Upward Movement (1): {target_counts.get(1, 0)*100:.2f}%")

plt.figure(figsize=(6, 4))
sns.barplot(x=['Down / Flat (0)', 'Up (1)'], y=[target_counts.get(0, 0), target_counts.get(1, 0)], palette='Blues_d')
plt.title('Directional Target Class Balance across S&P 500', fontweight='bold')
plt.ylabel('Proportion')
plt.ylim(0, 1.0)
plt.tight_layout()
plt.show()
```
